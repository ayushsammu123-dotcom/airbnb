# 01. Airbnb Data Cleaning & Validation Pipeline

This notebook demonstrates the end-to-end data cleaning, handling of missing values, duplicate removal, coordinate bounding, and price outlier treatment for Airbnb listing data.


```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import pandas as pd
import numpy as np
from src.data_cleaning import DataCleaner, clean_data
from src.utils import PROJECT_ROOT
```

## 1. Load Raw Dataset


```python
raw_path = project_root / 'data' / 'raw' / 'synthetic_airbnb.csv'
raw_df = pd.read_csv(raw_path)
print(f'Raw dataset loaded: {len(raw_df):,} rows, {raw_df.shape[1]} columns')
raw_df.head()
```

## 2. Inspect Raw Data Quality Issues
Check missing values, duplicate IDs, and price anomalies.


```python
print('--- Missing Values by Column ---')
print(raw_df.isnull().sum()[raw_df.isnull().sum() > 0])

print('\n--- Duplicates Count ---')
print(f'Duplicate listing IDs: {raw_df["listing_id"].duplicated().sum():,}')

print('\n--- Price Anomalies ---')
print(f'Negative / Zero prices: {(raw_df["price"] <= 0).sum():,}')
print(f'Extreme prices (> $1000): {(raw_df["price"] > 1000).sum():,}')
```

## 3. Execute Modular Cleaning Pipeline


```python
cleaner = DataCleaner(raw_df)
cleaned_df = cleaner.clean()
report = cleaner.generate_report()

print(f'Cleaned dataset: {len(cleaned_df):,} rows')
print('Cleaning Report Summary:')
for k, v in report.items():
    if k != 'steps':
        print(f'  {k}: {v}')
```

## 4. Save Processed Clean Dataset


```python
out_path = project_root / 'data' / 'processed' / 'airbnb_cleaned.csv'
out_path.parent.mkdir(parents=True, exist_ok=True)
cleaned_df.to_csv(out_path, index=False)
print(f'Cleaned dataset saved to {out_path}')
```

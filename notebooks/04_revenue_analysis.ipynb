# 04. Revenue Estimation & Host Portfolio Analytics

Detailed analysis of estimated annual and monthly revenue across neighbourhoods, room types, and host portfolios.


```python
import sys
from pathlib import Path

project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.feature_engineering import engineer_features
from src.analysis import compute_host_stats, get_top_listings

df = pd.read_csv(project_root / 'data' / 'processed' / 'airbnb_cleaned.csv')
df = engineer_features(df)
```

## 1. Top Revenue-Generating Neighbourhoods


```python
rev_neigh = df.groupby('neighbourhood')['estimated_annual_revenue'].agg(['sum', 'mean', 'count']).sort_values('sum', ascending=False)
rev_neigh.columns = ['Total Revenue', 'Avg Revenue per Listing', 'Listings']
print(rev_neigh.head(15))
```

## 2. Top Revenue-Generating Hosts


```python
top_hosts = compute_host_stats(df)
print(top_hosts.sort_values('total_estimated_revenue', ascending=False).head(10))
```

## 3. Revenue by Room Type & Host Category


```python
print(df.groupby(['room_type', 'host_category'])['estimated_annual_revenue'].agg(['mean', 'median', 'count']))
```

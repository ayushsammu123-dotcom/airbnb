# 02. Exploratory Data Analysis (EDA)

Comprehensive exploration of listing distributions, prices by neighbourhood and room type, availability patterns, and host metrics.


```python
import sys
from pathlib import Path

project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from src.feature_engineering import engineer_features

# Load cleaned data and engineer features
df = pd.read_csv(project_root / 'data' / 'processed' / 'airbnb_cleaned.csv')
df = engineer_features(df)
print(f'Dataset loaded: {len(df):,} records')
```

## 1. Price Distribution & Boxplots


```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df['price'], bins=40, kde=True, color='#FF385C')
plt.title('Nightly Price Distribution ($)')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='room_type', y='price', palette='Set2')
plt.title('Price by Room Type')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 2. Neighbourhood Listing Count & Average Price


```python
neigh_summary = df.groupby('neighbourhood').agg(count=('listing_id', 'count'), avg_price=('price', 'mean')).sort_values('count', ascending=False)
print(neigh_summary.head(15))
```
